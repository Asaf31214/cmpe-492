```python
from os import path

import polars as pl
from IPython.display import display
```

```python
tabular_data_path = path.join("..", "..", "data", "tabular")
source_data_xlsx = path.join(tabular_data_path, "pediatric-pcnl.xlsx")
source_data_csv = path.join(tabular_data_path, "pediatric-pcnl.csv")
processed_csv = path.join(tabular_data_path, "processed.csv")
```

```python
source_data = pl.read_excel(source_data_xlsx)
source_data.write_csv(source_data_csv)
```

```python
# ============================================
# 1. BASIC DATA OVERVIEW
# ============================================
print("=" * 60)
print("1. DATASET SHAPE & COLUMNS")
print("=" * 60)
print(f"Shape: {source_data.shape}")
print(f"\nColumns ({len(source_data.columns)}):")
for i, col in enumerate(source_data.columns, 1):
    print(f"{i:3}. {col}")

# ============================================
# 2. DATA TYPES & MISSING VALUES
# ============================================
print("\n" + "=" * 60)
print("2. DATA TYPES & MISSING VALUE ANALYSIS")
print("=" * 60)

missing_info = source_data.select([pl.col(c).alias(c) for c in source_data.columns]).select(
    [pl.col(c).null_count().alias(f"{c}_null") for c in source_data.columns]
)

missing_pct = source_data.select(
    [(pl.col(c).null_count() / len(source_data) * 100).alias(f"{c}_pct") for c in source_data.columns]
)

print("\nMissing values (>10% highlighted):")
for col in source_data.columns:
    null_count = source_data[col].null_count()
    null_pct = null_count / len(source_data) * 100
    if null_pct > 10:
        print(f"⚠️  {col:40}: {null_count:4} ({null_pct:5.1f}%)")
    elif null_pct > 0:
        print(f"   {col:40}: {null_count:4} ({null_pct:5.1f}%)")

# ============================================
# 3. TARGET VARIABLE DISTRIBUTION
# ============================================
print("\n" + "=" * 60)
print("3. TARGET VARIABLE DISTRIBUTION")
print("=" * 60)

# Check both SONUC and SONUÇ-2
for target in ["SONUC", "SONUÇ-2"]:
    if target in source_data.columns:
        print(f"\n{target}:")
        dist = source_data[target].value_counts().sort("count", descending=True)
        print(dist)
        total = dist["count"].sum()
        for row in dist.iter_rows():
            print(f"  {row[0]}: {row[1]} ({row[1] / total * 100:.1f}%)")

# ============================================
# 4. GUY'S STONE SCORE DISTRIBUTION (KEY FEATURE)
# ============================================
print("\n" + "=" * 60)
print("4. GUY'S STONE SCORE DISTRIBUTION")
print("=" * 60)

if "GUY SCORE" in source_data.columns:
    guy_dist = source_data["GUY SCORE"].value_counts().sort("GUY SCORE")
    print(guy_dist)

    total = guy_dist["count"].sum()
    print("\nPercentage by Grade:")
    for row in guy_dist.iter_rows():
        print(f"  Grade {row[0]}: {row[1]} ({row[1] / total * 100:.1f}%)")

    # Compare with literature (Kırlı et al. 2020)
    print("\n📚 Literature Comparison (Kırlı et al. 2020 - Pediatric PCNL):")
    print("  Grade 1: 24% | Grade 2: 46% | Grade 3: 16% | Grade 4: 14%")
else:
    print("⚠️  GUY SCORE column not found!")

# ============================================
# 5. STONE-FREE RATE BY GUY SCORE
# ============================================
print("\n" + "=" * 60)
print("5. STONE-FREE RATE BY GUY SCORE (FIXED)")
print("=" * 60)

if "GUY SCORE" in source_data.columns and "SONUÇ-2" in source_data.columns:
    # Filter out nulls and cast to proper types
    df_temp = source_data.filter(pl.col("GUY SCORE").is_not_null() & pl.col("SONUÇ-2").is_not_null()).with_columns(
        [pl.col("GUY SCORE").cast(pl.Int64, strict=False), pl.col("SONUÇ-2").cast(pl.Int64, strict=False)]
    )

    sfr_by_guy = (
        df_temp.group_by("GUY SCORE")
        .agg(
            [
                pl.col("SONUÇ-2").count().alias("total"),
                (pl.col("SONUÇ-2") == 1).sum().alias("stone_free"),
                ((pl.col("SONUÇ-2") == 1).sum() / pl.col("SONUÇ-2").count() * 100).alias("sfr_pct"),
            ]
        )
        .sort("GUY SCORE")
    )

    print(sfr_by_guy)

    print("\n📚 Literature Comparison (Kırlı et al. 2020):")
    print("  Grade 1: 89% | Grade 2: 78% | Grade 3: 75% | Grade 4: 57%")
    print("  Overall: 77%")
else:
    print("⚠️  Cannot calculate - missing GUY SCORE or SONUÇ-2")

# ============================================
# 6. KEY FEATURE DISTRIBUTIONS
# ============================================
print("\n" + "=" * 60)
print("6. KEY FEATURE DISTRIBUTIONS")
print("=" * 60)

key_features = [
    "YAŞ",
    "CİNSİYET",
    "LOKALİZASYON",
    "TOPLAM TAŞ YÜKÜ (CM2)",
    "ASA SKORU",
    "PEROPKOMPLİKASYON",
    "POSTOPKOMPLİKASYON",
    "CLAVİEN SİSTEM",
    "TRANSFÜZYON",
    "ATEŞ",
    "GUY SCORE",
]

for col in key_features:
    if col in source_data.columns:
        print(f"\n{col}:")
        null_count = source_data[col].null_count()
        print(f"  Nulls: {null_count} ({null_count / len(source_data) * 100:.1f}%)")

        # Cast to numeric if possible
        col_data = source_data[col].cast(pl.Int64, strict=False).drop_nulls()

        if len(col_data) > 0:
            unique_vals = col_data.unique()
            if len(unique_vals) <= 10:
                dist = col_data.value_counts().sort(col)
                print(f"  Distribution:\n{dist}")
            else:
                print(f"  Min: {col_data.min()}, Max: {col_data.max()}, Median: {col_data.median()}")

# ============================================
# 7. COMPLICATION RATES
# ============================================
print("\n" + "=" * 60)
print("7. COMPLICATION RATES")
print("=" * 60)

for comp_col in ["PEROPKOMPLİKASYON", "POSTOPKOMPLİKASYON", "CLAVİEN SİSTEM", "TRANSFÜZYON", "ATEŞ"]:
    if comp_col in source_data.columns:
        valid_data = source_data.filter(pl.col(comp_col).is_not_null()).with_columns(
            [pl.col(comp_col).cast(pl.Int64, strict=False)]
        )

        if len(valid_data) > 0:
            unique_vals = valid_data[comp_col].unique()
            if len(unique_vals) <= 5:
                dist = valid_data[comp_col].value_counts().sort(comp_col)
                print(f"\n{comp_col} (n={len(valid_data)}):")
                print(dist)

                # Calculate complication rate if binary (0/1)
                if 0 in unique_vals and 1 in unique_vals:
                    comp_rate = (valid_data[comp_col] == 1).sum() / len(valid_data) * 100
                    print(f"  ⚠️  Complication Rate: {comp_rate:.1f}%")
                    print("  📚 Literature (Kırlı et al. 2020): 17%")

# ============================================
# 8. CORRELATION WITH TARGET
# ============================================
print("\n" + "=" * 60)
print("8. CORRELATION WITH TARGET (SONUÇ-2)")
print("=" * 60)

numeric_cols = [
    "YAŞ",
    "ASA SKORU",
    "TOPLAM TAŞ YÜKÜ (CM2)",
    "GUY SCORE",
    "OP SÜRESİ",
    "SKOPI SURE",
    "NEFROSTOMI SURESI",
    "PCNL SEANS SAYISI",
    "hastanede kalis",
    "TRAKT SAYISI",
]

# Filter to numeric columns that exist
available_numeric = [c for c in numeric_cols if c in source_data.columns]

if "SONUÇ-2" in source_data.columns and len(available_numeric) > 0:
    # Cast everything to numeric
    corr_data = (
        source_data.select(available_numeric + ["SONUÇ-2"])
        .with_columns([pl.col(c).cast(pl.Float64, strict=False) for c in available_numeric + ["SONUÇ-2"]])
        .drop_nulls()
    )

    if len(corr_data) > 10:
        print(f"\nCorrelation with SONUÇ-2 (n={len(corr_data)}):")
        correlations = []
        for col in available_numeric:
            if col != "SONUÇ-2":
                try:
                    corr = corr_data.select(pl.col(col).corr(pl.col("SONUÇ-2"))).item()
                    if corr is not None and not (corr != corr):  # Check for NaN
                        correlations.append((col, corr))
                except:
                    pass

        correlations.sort(key=lambda x: abs(x[1]), reverse=True)
        for col, corr in correlations[:15]:
            marker = "🔴" if abs(corr) > 0.3 else "🟡" if abs(corr) > 0.1 else "⚪"
            print(f"  {marker} {col:35}: {corr:+.3f}")

# ============================================
# 9. FEATURE CANDIDATES BY MISSINGNESS
# ============================================
print("\n" + "=" * 60)
print("9. FEATURE CANDIDATES BY MISSINGNESS")
print("=" * 60)

feature_categories = {
    "✅ KEEP (<10% missing)": [],
    "⚠️  CONSIDER (10-30% missing)": [],
    "❌ DROP (>30% missing)": [],
    "🎯 TARGET/KEY": [],
}

for col in source_data.columns:
    null_pct = source_data[col].null_count() / len(source_data) * 100

    if col in ["SONUC", "SONUÇ-2", "CLAVİEN SİSTEM"]:
        feature_categories["🎯 TARGET/KEY"].append((col, null_pct))
    elif col in ["GUY SCORE", "TOPLAM TAŞ YÜKÜ (CM2)", "LOKALİZASYON"]:
        feature_categories["🎯 TARGET/KEY"].append((col, null_pct))
    elif null_pct < 10:
        feature_categories["✅ KEEP (<10% missing)"].append((col, null_pct))
    elif null_pct < 30:
        feature_categories["⚠️  CONSIDER (10-30% missing)"].append((col, null_pct))
    else:
        feature_categories["❌ DROP (>30% missing)"].append((col, null_pct))

for category, cols in feature_categories.items():
    print(f"\n{category} ({len(cols)} columns):")
    for col, pct in sorted(cols, key=lambda x: x[1])[:15]:
        print(f"   {col:45}: {pct:5.1f}%")

# ============================================
# Age Distribution
# ============================================
print("\n" + "=" * 60)
print("AGE DISTRIBUTION (Fixed)")
print("=" * 60)

if "YAŞ" in source_data.columns:
    age_data = source_data["YAŞ"].cast(pl.Int64, strict=False).drop_nulls()

    # Simple age bins
    age_groups = age_data.cut([2, 6, 12, 17], include_breaks=True)
    print(age_groups.value_counts().sort("break_point"))

    # Count by pediatric categories
    infant = (age_data <= 2).sum()
    child = ((age_data > 2) & (age_data <= 12)).sum()
    adolescent = ((age_data > 12) & (age_data <= 17)).sum()
    adult = (age_data > 17).sum()

    print(f"\n  Infant (0-2): {infant} ({infant / len(age_data) * 100:.1f}%)")
    print(f"  Child (3-12): {child} ({child / len(age_data) * 100:.1f}%)")
    print(f"  Adolescent (13-17): {adolescent} ({adolescent / len(age_data) * 100:.1f}%)")
    print(f"  Adult (>17): {adult} ({adult / len(age_data) * 100:.1f}%)")

print("\n" + "=" * 60)
print("CORRELATION WITH TARGET")
print("=" * 60)

# Create correlation dataframe with proper casting
corr_cols = ["YAŞ", "ASA SKORU", "GUY SCORE", "LOKALİZASYON", "TOPLAM TAŞ YÜKÜ (CM2)", "TRAKT SAYISI"]

corr_df = (
    source_data.select(corr_cols + ["SONUÇ-2"])
    .with_columns(
        [
            pl.col("SONUÇ-2").cast(pl.Int64, strict=False),
            pl.col("GUY SCORE").cast(pl.Float64, strict=False),
            pl.col("LOKALİZASYON").cast(pl.Float64, strict=False),
            pl.col("TOPLAM TAŞ YÜKÜ (CM2)").cast(pl.Float64, strict=False),
            pl.col("YAŞ").cast(pl.Float64, strict=False),
            pl.col("ASA SKORU").cast(pl.Float64, strict=False),
            pl.col("TRAKT SAYISI").cast(pl.Float64, strict=False),
        ]
    )
    .drop_nulls()
)

print(f"Sample size for correlation: {len(corr_df)}")

if len(corr_df) > 50:
    print("\nCorrelation with SONUÇ-2 (1=Stone Free, 2=Residual):")
    print("(Negative = higher value predicts LOWER success)")
    for col in corr_cols:
        if col != "SONUÇ-2":
            try:
                corr = corr_df.select(pl.col(col).corr(pl.col("SONUÇ-2"))).item()
                if corr is not None:
                    marker = "🔴" if abs(corr) > 0.2 else "🟡" if abs(corr) > 0.1 else "⚪"
                    print(f"  {marker} {col:35}: {corr:+.3f}")
            except Exception as e:
                print(f"  ⚠️  {col:35}: Error - {str(e)[:50]}")
```

```python
age_data = source_data["YAŞ"].cast(pl.Int64, strict=False).drop_nulls()

print(f"Age Statistics (n={len(age_data)}):")
print(f"  Min: {age_data.min()}, Max: {age_data.max()}")
print(f"  Mean: {age_data.mean():.1f}, Median: {age_data.median():.1f}")

infant = (age_data <= 2).sum()
child = ((age_data > 2) & (age_data <= 12)).sum()
adolescent = ((age_data > 12) & (age_data <= 17)).sum()
adult = (age_data > 17).sum()
total = len(age_data)

print(f"\n  Infant (0-2): {infant} ({infant / total * 100:.1f}%)")
print(f"  Child (3-12): {child} ({child / total * 100:.1f}%)")
print(f"  Adolescent (13-17): {adolescent} ({adolescent / total * 100:.1f}%)")
print(f"  Adult (>17): {adult} ({adult / total * 100:.1f}%)")

corr_cols = ["YAŞ", "ASA SKORU", "GUY SCORE", "LOKALİZASYON", "TOPLAM TAŞ YÜKÜ (CM2)", "TRAKT SAYISI"]

corr_df = (
    source_data.select(corr_cols + ["SONUÇ-2"])
    .with_columns(
        [
            pl.col("SONUÇ-2").cast(pl.Int64, strict=False),
            pl.col("GUY SCORE").cast(pl.Float64, strict=False),
            pl.col("LOKALİZASYON").cast(pl.Float64, strict=False),
            pl.col("TOPLAM TAŞ YÜKÜ (CM2)").cast(pl.Float64, strict=False),
            pl.col("YAŞ").cast(pl.Float64, strict=False),
            pl.col("ASA SKORU").cast(pl.Float64, strict=False),
            pl.col("TRAKT SAYISI").cast(pl.Float64, strict=False),
        ]
    )
    .drop_nulls()
)

print(f"Correlation with SONUÇ-2 (n={len(corr_df)}):")
print("(Negative = higher value predicts LOWER success)\n")

for col in corr_cols:
    try:
        corr = corr_df.select(pl.col(col).corr(pl.col("SONUÇ-2"))).item()
        if corr is not None:
            print(f"  {col:35}: {corr:+.3f}")
    except:
        pass

summary_rows = []
for col in source_data.columns:
    summary_rows.append(
        {
            "column": col,
            "null_count": source_data[col].null_count(),
            "null_pct": source_data[col].null_count() / len(source_data) * 100,
            "dtype": str(source_data[col].dtype),
            "unique_count": source_data[col].n_unique(),
        }
    )

summary_df = pl.DataFrame(summary_rows)
summary_df.write_csv(path.join(tabular_data_path, "column_summary.csv"))

print(f"Column summary saved: {len(summary_df)} columns")

guy_sfr = (
    source_data.filter(pl.col("GUY SCORE").is_not_null() & pl.col("SONUÇ-2").is_not_null())
    .with_columns([pl.col("GUY SCORE").cast(pl.Int64, strict=False), pl.col("SONUÇ-2").cast(pl.Int64, strict=False)])
    .group_by("GUY SCORE")
    .agg(
        [
            pl.col("SONUÇ-2").count().alias("total"),
            (pl.col("SONUÇ-2") == 1).sum().alias("stone_free"),
            ((pl.col("SONUÇ-2") == 1).sum() / pl.col("SONUÇ-2").count() * 100).alias("sfr_pct"),
        ]
    )
    .sort("GUY SCORE")
)

guy_sfr.write_csv(path.join(tabular_data_path, "guy_score_sfr.csv"))

print("Guy's Score SFR saved")
display(guy_sfr)
```

```python
drop_cols = [
    "PTS NO",
    "UNITE",
    "AD",
    "SOYAD",
    "OP TARİH",
    "TABURCU TARİHİ",
    "AĞIRLIK",
    "BOY",
    "IGNEDEN GELEN IDRAR RENGI",
    "TAS OPASITE",
    "NEFROSKOP CAPI",
    "SKOPI SURE",
    "Perioperatif kullanılan antibiyotik",
    "POSTOP HCT",
    "POSTOP KREATİNİN",
    "NEFROSTOMI SURESI",
    "TAS ANALIZI",
    "TAŞOPASITE",
    "PREOPHEMATOKRİT",
    "SHEAT BOYU",
    "OZETSHEATBOYU",
    "GEÇ. CERRAHİ ÖZET",
    "ozetIKAB",
    "ÖZET KOMP.",
]

feature_cols = [
    "YAŞ",
    "CİNSİYET",
    "TARAF",
    "ASA SKORU",
    "BT",
    "LOKALİZASYON",
    "GİRİŞ REHBERİ",
    "GİRİŞ KALİSİ",
    "GIRIS LOK",
    "DİLATASYON",
    "TRAKT SAYISI",
    "KIRMA YÖNTEMİ",
    "POSTOP NEFROSTOMİ",
    "POSTOP ÜRETER KATETERİ",
    "PEROPKOMPLİKASYON",
    "TRANSFÜZYON",
    "ATEŞ",
    "POSTOPKOMPLİKASYON",
    "PCNL SEANS SAYISI",
    "EK TEDAVİ",
    "GUY SCORE",
    "SATAVA S.",
    "TOPLAM TAŞ YÜKÜ (CM2)",
    "OP SÜRESİ",
    "hastanede kalis",
    "ÖZGEÇMİŞ",
    "GEÇİRİLMİŞ CERRAHİ",
    "TAS ANAMNEZI",
    "İKAB",
    "KREATİNİN",
    "ESZAMANLI",
    "SOLİTER BB",
    "RENAL ANOMALİ",
    "EK RENAL HASTALIK",
    "VUCUT ANOMALI",
]

target_col = "SONUÇ-2"

df_ml = source_data.select(feature_cols + [target_col])

numeric_cols = [
    "YAŞ",
    "ASA SKORU",
    "TOPLAM TAŞ YÜKÜ (CM2)",
    "OP SÜRESİ",
    "TRAKT SAYISI",
    "PCNL SEANS SAYISI",
    "hastanede kalis",
    "KREATİNİN",
    "GUY SCORE",
    "SATAVA S.",
]

categorical_cols = [c for c in feature_cols if c not in numeric_cols]

for col in numeric_cols:
    if col in df_ml.columns:
        median_val = df_ml[col].median()
        if median_val is None:
            median_val = 0
        df_ml = df_ml.with_columns(pl.col(col).fill_null(median_val))

for col in categorical_cols:
    if col in df_ml.columns:
        mode_val = df_ml[col].mode()
        if len(mode_val) > 0:
            fill_val = mode_val[0]
        else:
            fill_val = 0
        df_ml = df_ml.with_columns(pl.col(col).fill_null(fill_val))

df_ml = df_ml.with_columns([pl.col(c).cast(pl.Int64, strict=False) for c in df_ml.columns])

df_ml = df_ml.drop_nulls(subset=[target_col])

print(f"Final ML dataset shape: {df_ml.shape}")
print("Target distribution:")
display(df_ml[target_col].value_counts())
print("\nMissing values after preprocessing:")
display(df_ml.null_count())

df_ml.write_csv(processed_csv)
print(f"\nSaved to: {processed_csv}")
```
